Step1: Display the Dataset

```python
#Import Pandas Library
import pandas as pd
# Create a dictionary containing employee data
data={
    "Emp_ID":[101,102,103,104,105,106,107,108],
    "Name":["Rahul","Priya","Amit","Riya","Karan","Neha","Vivek","Pooja"],
    "Department":["IT","HR","IT","Finance","HR","Finance","IT","HR"],
    "City":["Surat","Navsari","Surat","Valsad","Surat","Navsari","Valsad","Surat"],
    "Salary":[45000,38000,52000,48000,40000,51000,60000,42000],
    "Experience":[2,1,5,4,2,6,8,3]
}
# Convert dictionary into a Pandas DataFrame
df=pd.DataFrame(data)
```

```python
# Display the DataFrame
df
```

Step 2: Filter Data

Display Employees From IT Department

```python
# Display only the Employees whose Department is "IT"
df[df["Department"] == "IT"]
```

Display Employees with Salary Greater Than 50,000

```python
#Display the Employees whose salary is greater than 50,000
df[df["Salary"]>50000]
```

Display Employee From Surat

```python
#Display the Employees who are from "Surat"
df[df["City"]=="Surat"]
```

Apply Multiple Conditions

```python
#Display the Employees whose Department is "IT"
#Display the Employees whose Salary is greater than 50,00
df[(df["Department"]=="IT") & (df["Salary"]>50000)]
```

Step 3: Sort Data

Sort Salary in Ascending Order

```python
#Sort the DataFrame by Salary in ascending order
df.sort_values(by="Salary")
```

Sort Salary in Descending Order

```python
#Sort the DataFrame by Salary in descending order
df.sort_values(by="Salary",ascending=False)
```

Sort by Department and Salary

```python
#Sort the DataFrame by Department in ascending order
#Sort the DataFrame by  Salary in decending order
df.sort_values(by=["Department","Salary"],ascending=[True,False])
```

Step 4: Perform Aggregation


Calculate Total Salary

```python
#Calculate the total sum of all employee salaries
df["Salary"].sum()
```

Calculate Average Salary

```python
#Calculate the average salary of all employees
df["Salary"].mean()
```

Find Highest Salary

```python
#Find the Highest salary from the Salary column
df["Salary"].max()
```

Find Lowest Salary

```python
#Find the Lowest salary from the Salary column
df["Salary"].min()
```

Count Employees

```python
#Count the number of non-empty values in the Emp_ID column
df["Emp_ID"].count()
```

Step5: Perform GroupBy Operations

Average Salary Department-wise

```python
#Calculate the Average Salary for each Department
df.groupby("Department")["Salary"].mean()
```

Minimum Salary Department-wise

```python
#Calculate the Minimum Salary for each Department
df.groupby("Department")["Salary"].min()
```

Count Total Employees Department-wise

```python
#Group the data by "Department"
#Count the number of Emp_ID values in each department
df.groupby("Department")["Emp_ID"].count()
```

Group by City

```python
#Group the data by "City" and Calculate the Average Salary for each City
df.groupby("City")["Salary"].mean()
```

Total Salary Deparetment-wise

```python
#Group the data by "Department"
#Calculate the Total Salary for each Department
df.groupby("Department")["Salary"].sum()
```

Step 6: Multiple Aggregation Functions

```python
#Group the data by "Department"
#Calculate multiple statistics for the Salary column
df.groupby("Department")["Salary"].agg(["count","sum","mean","min","max"])
```

Step 7: Save the Result

```python
#Save the DataFrame as a CSV file named "Employee_Data.csv"
df.to_csv("Employee_Data.csv",index=False)
print("Dataset Saved Successfully.")
```
