AIM: Perform data inspection, cleaning, handling missing values, duplicate removal, and data transformation using Pandas.


Step 1: Import Pandas Library

```python
#Import Pandas library
import pandas as pd
```

Step 2: Create the Dataset

```python
#Create Sample Data
data = {
    "ID": [101, 102, 103, 104, 105, 106, 107,108],
    "Name": ["Rahul","Priya","Amit",None,"Riya","Ajay","Naresh","Jenil"],
    "Age": [20,21,None,20,20,22,23,28],
    "City": ["Surat","Navsari","Surat","Bharuch","Bharuch","Navsari","Surat","Ahemedabad"],
    "Marks": [85,None,78,88,88,81,90,93]
}

df = pd.DataFrame(data)
```

Step 3: Inspect the Dataset

```python
#Display the First Five Rows of the Dataset
df.head()
```

```python
#Display Last Five Rows of Dataset
df.tail()
```

```python
#Display Detailed Information about the Dataset
df.info()
```

```python
#Generate Statistical Information for Numerical Columns
df.describe()
```

```python
#Display Dimension of the Dataset
df.shape
```

```python
#Display the names of all Columns in the DataFrame
df.columns
```

```python
#Display the Data Type of each Column
df.dtypes
```

Step 4: Check Missing Values

```python
#Check for missing(null) values in every column
df.isnull().sum()
```

Step 5: Fill Missing Marks

```python
#Replace missing values in the marks column
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

#Display the updated DataFrame
df
```

Fill Missing Age

```python
#Replace missing values in the Age Column
df["Age"] = df["Age"].fillna(df["Age"].mean())

#Display the updated DataFrame
df
```

Step 6: Detect Duplicate Records

```python
#Check whether each row is duplicated
df.duplicated()
```

```python
#Count the total number of duplicate rows
df.duplicated().sum()
```

Step 7: Remove Duplicate Records

```python
#Remove duplicate rows from the DataFrame
df = df.drop_duplicates()

#Display the cleaned DatFrame
df
```

Step 8: Rename a Column

```python
#Rename the Column "Marks" to "Score"
df.rename(columns={"Marks":"Score"}, inplace=True)

#Display the updated DataFrame
df
```

Step 9: Change Data Type

```python
#Convert the Age Column from float to integer
df["Age"] = df["Age"].astype(int)

#Display the updated Data Types
df.dtypes
```

Step 10: Create a New Column

```python
#Create a new column named "Final Score"
df["Final Score"] = df["Score"] + 5

#Display the updated DataFrame
df
```

Step 11:  Convert Names to Uppercase

```python
#Convert all Student Names to Uppercase
df["Name"] = df["Name"].str.upper()

#Display the updated DataFrame
df
```

Step 12: Convert Text to Lowercase

```python
#Convert all City Names to Lowercase
df["City"] = df["City"].str.lower()

#Display the updated DataFrame
df
```

Step 13: Remove Extra Spaces

```python
#Remove leading and trailing spaces from City names
df["City"] = df["City"].str.strip()

#Display the updated DataFrame
df
```

Step 14: Sort the Dataset

```python
#Sort the DataFrame based on the "Final Score" column
df = df.sort_values(by="Final Score", ascending=False)
```

Step 15: Save the Cleaned Dataset

```python
#Save the cleaned DataFarame as a new CSV file
df.to_csv("cleaned_students.csv", index=False)

#Display a confirmation message
print("Dataset Saved Successfilly.")
```
