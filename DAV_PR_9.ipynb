Step 2: Import libraries

```python
import dash
from dash import dcc, html, Input, Output
import plotly.express as px
import pandas as pd
```

Step 3: Create sample dataset

```python
# Create social media dataset
data = {
    "Platform": ["Instagram", "Facebook", "YouTube", "Instagram",
                 "Facebook", "YouTube", "Instagram", "Facebook",
                 "YouTube", "Instagram"],

    "Month": ["Jan", "Jan", "Jan", "Feb", "Feb",
              "Feb", "Mar", "Mar", "Mar", "Apr"],

    "Likes": [500, 300, 700, 600, 400, 800, 750, 450, 900, 850],

    "Comments": [100, 80, 150, 120, 90, 180, 140, 110, 200, 160],

    "Shares": [50, 40, 100, 70, 60, 120, 90, 75, 150, 110],

    "Views": [5000, 3000, 10000, 6000, 4000,
              12000, 7500, 5000, 15000, 9000]
}

df = pd.DataFrame(data)

# Display dataset
display(df)
```

Step 4: Data Inspection



```python
# Display first five records
print("First Five Records:")
display(df.head())

# Display dataset information
print("Dataset Information:")
df.info()

# Display statistical summary
print("Statistical Summary:")
display(df.describe())
```

Step 5: Check missing values

```python
# Check missing values in each column
print("Missing Values:")
print(df.isnull().sum())
```

Step 6: Handle missing values

```python
# Fill missing numerical values with median
df["Likes"] = df["Likes"].fillna(df["Likes"].median())

df["Comments"] = df["Comments"].fillna(df["Comments"].median())

df["Shares"] = df["Shares"].fillna(df["Shares"].median())

df["Views"] = df["Views"].fillna(df["Views"].median())

# Fill missing categorical values with mode
df["Platform"] = df["Platform"].fillna(df["Platform"].mode()[0])

df["Month"] = df["Month"].fillna(df["Month"].mode()[0])

print("Missing values handled successfully.")

# Check missing values again
print(df.isnull().sum())
```

Step 7: Check and remove duplicate data

```python
# Check duplicate records
print("Duplicate Records:", df.duplicated().sum())

# Remove duplicate records
df = df.drop_duplicates()

print("Duplicate records removed.")

# Display updated dataset
display(df)
```

Step 8: Data transformation

```python
# Calculate total engagement
df["Engagement"] = (
    df["Likes"] + df["Comments"] + df["Shares"]
)

# Calculate engagement rate
df["Engagement_Rate"] = (
    df["Engagement"] / df["Views"] * 100
)

# Display transformed data
display(df)
```

Step 9: Create dashboard layout

```python
# Create Dash application
app = dash.Dash(__name__)

app.layout = html.Div([
    html.H1(
        "Social Media Analysis Dashboard",
        style={"textAlign": "center", "color": "blue"}
    ),

    html.H3("Select Platform:"),

    dcc.Dropdown(
        id="platform",
        options=[
            {"label": "All Platforms", "value": "All"},
            {"label": "Instagram", "value": "Instagram"},
            {"label": "Facebook", "value": "Facebook"},
            {"label": "YouTube", "value": "YouTube"}
        ],
        value="All",
        clearable=False
    ),

    html.Br(),

    html.Div(id="summary"),

    dcc.Graph(id="likes-chart"),
    dcc.Graph(id="engagement-chart"),
    dcc.Graph(id="views-chart"),
    dcc.Graph(id="rate-chart")
])
```

Step 10: Add interactive charts

```python
@app.callback(
    [Output("summary", "children"),
     Output("likes-chart", "figure"),
     Output("engagement-chart", "figure"),
     Output("views-chart", "figure"),
     Output("rate-chart", "figure")],

    Input("platform", "value")
)

def update_dashboard(selected_platform):

    # Filter selected platform
    if selected_platform == "All":
        filtered_df = df
    else:
        filtered_df = df[df["Platform"] == selected_platform]

    # Calculate summary
    total_likes = filtered_df["Likes"].sum()
    total_comments = filtered_df["Comments"].sum()
    total_views = filtered_df["Views"].sum()

    summary = html.Div([
        html.H3(f"Total Likes: {total_likes:,}"),
        html.H3(f"Total Comments: {total_comments:,}"),
        html.H3(f"Total Views: {total_views:,}")
    ], style={
        "backgroundColor": "#e6f0ff",
        "padding": "15px",
        "textAlign": "center"
    })

    # Likes chart
    likes_fig = px.bar(
        filtered_df,
        x="Month",
        y="Likes",
        color="Platform",
        title="Monthly Likes Analysis"
    )

    # Engagement chart
    engagement_data = filtered_df.groupby("Platform")[
        ["Likes", "Comments", "Shares"]
    ].sum().reset_index()

    engagement_fig = px.bar(
        engagement_data,
        x="Platform",
        y=["Likes", "Comments", "Shares"],
        barmode="group",
        title="Social Media Engagement Analysis"
    )

    # Views chart
    views_data = filtered_df.groupby("Platform")[
        "Views"
    ].sum().reset_index()

    views_fig = px.pie(
        views_data,
        names="Platform",
        values="Views",
        title="Views by Platform"
    )

    # Engagement rate chart
    rate_data = filtered_df.groupby("Platform")[
        "Engagement_Rate"
    ].mean().reset_index()

    rate_fig = px.bar(
        rate_data,
        x="Platform",
        y="Engagement_Rate",
        title="Average Engagement Rate"
    )

    return summary, likes_fig, engagement_fig, views_fig, rate_fig
```
