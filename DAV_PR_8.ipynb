Step 1: Import Required Libraris

```python
# Import pandas library
import pandas as pd

# Import Plotly Express for creating simple and interactive charts
import plotly.express as px

# Import Plotly Graph Objects for creating more customized charts
import plotly.graph_objects as go
```

Step 2: Create a Common Dataset

```python
data ={
    "Month": ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
              'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
"Product":['Laptop', 'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet','Laptop', 'Mobile', 'Tablet','Laptop', 'Mobile', 'Tablet'],
    "Region":['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
"Sales": [120000, 135000, 150000, 145000, 160000, 175000,
                180000, 170000, 190000, 210000, 230000, 250000],
    "Units_Sold":[12,30,25,14,35,30,18,38,32,21,45,40],
    "Profit":[18000, 27000, 30000, 22000, 32000, 35000, 28000, 34000, 38000, 33000, 46000, 50000]
    }

df = pd.DataFrame(data)

print(df)
```

Step 3: Interactive Line Chart Using Plotly Express

```python
import pandas as pd
import plotly.express as px

# Check column names
print(df.columns)

# Create interactive line chart using the correct 'Sales' column
fig = px.line(
    df,
    x="Month",
    y="Sales",
    title="Sales Revenue Throughout the Year",
    markers=True
)

# Axis labels
fig.update_xaxes(title_text="Month")
fig.update_yaxes(title_text="Sales")

# Display chart
fig.show()
```

Step 4: Improve the Hover Information

```python
# Create an interactive line chart using Plotly Express
fig = px.line(
    df,
    x="Month",
    y="Sales",
    markers=True,
    title="Sales Over Month",

    # Add additional information to the hover box
    hover_data=[
        "Product",
        "Region",
        "Units_Sold",
        "Profit"
    ]
)

# Add X-axis label
fig.update_xaxes(title="Month")

# Add Y-axis label
fig.update_yaxes(title="Sales")

# Display the interactive line chart
fig.show()
```

Step 5: Create an Interactive Bar Chart

```python
# Group the dataset according to Product
# Calculate the total Sales for each product
product_sales = df.groupby("Product", as_index=False)["Sales"].sum()

# Display the grouped data
print(product_sales)
```

```python
# Create an interactive bar chart
fig = px.bar(
    product_sales,
    x="Product",
    y="Sales",
    title="Total Sales by Product",
    text="Sales"
)

# Add X-axis label
fig.update_xaxes(title="Product")

# Add Y-axis label
fig.update_yaxes(title="Total Sales")

# Display the interactive bar chart
fig.show()
```

```python
# Create an interactive scatter plot
fig = px.scatter(
    df,
    x="Units_Sold",
    y="Sales",
    color="Product",
    hover_data=[
        "Region",
        "Profit"
    ],
    title="Relationship Between Units Sold and Sales"
)

# Add X-axis label
fig.update_xaxes(title="Units Sold")

# Add Y-axis label
fig.update_yaxes(title="Sales")

# Display the interactive scatter plot
fig.show()
```

Step 6: Create an Interactive Scatter Plot

```python
df = pd.DataFrame({
    "Study Hours": [1, 2, 3, 4, 5, 6, 7, 8, 9],
    "Exam Score":  [30, 42, 45, 58, 62, 70, 76, 85, 88]
})

fig = px.scatter(
    df,
    x="Study Hours",
    y="Exam Score",
    title="Study Hours vs Exam Scores",
    hover_data=["Study Hours", "Exam Score"],
    trendline="ols"
)

fig.show()
```

Step 7: Create an Interactive Scatter Plot with Colour

```python
# Sample scatter plot data
df = pd.DataFrame({
    "Study Hours": [1, 2, 3, 4, 5, 6, 7, 8, 9],
    "Exam Score": [30, 42, 45, 58, 62, 70, 76, 85, 88]
})

# Import LinearRegression and initialize the model
from sklearn.linear_model import LinearRegression
model = LinearRegression()

# Regression-based colour variable
X = df[["Study Hours"]]
y = df["Exam Score"]
model.fit(X, y)

df["Predicted Score"] = model.predict(X)
df["Regression"] = df.apply(
    lambda row: "Above trend"
    if row["Exam Score"] >= row["Predicted Score"]
    else "Below trend",
    axis=1
)

fig1 = px.scatter(
    df,
    x="Study Hours",
    y="Exam Score",
    color="Regression",
    hover_data=["Predicted Score"],
    trendline="ols",
    title="Interactive Scatter Plot with Regression Colours"
)
fig1.show()
```

Step 8: Create an Interactive Pie Chart

```python
#To visualize regional sales contribution using an interactive Pie
# Regional sales pie chart
sales = pd.DataFrame({
    "Region": ["North", "South", "East", "West"],
    "Sales": [35000, 28000, 20000, 17000]
})

fig2 = px.pie(
    sales,
    names="Region",
    values="Sales",
    title="Regional Sales Contribution",
    hole=0.3
)

fig2.update_traces(
    textinfo="label+percent",
    hovertemplate=(
        "Region: %{label}<br>"
        "Sales: %{value:,.0f}<br>"
        "Share: %{percent}<extra></extra>"
    )
)
fig2.show()
```
